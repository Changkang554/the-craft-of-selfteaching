```python
import random
n = 3
a_list = [random.randrange(65, 91) for i in range(n)]
b_list = [chr(random.randrange(65, 91)) for i in range(n)]
print(a_list)
c_list = a_list + b_list +a_list * 2
print(c_list)

# list.append() 在列表后面加一个元素
c_list.append('100')
print(c_list)

# list.clear()清空list
print()
print(a_list)
a_list.clear()
print(a_list)

# list.copy()拷贝一个列表
print()
d_list = c_list.copy()
print(d_list)
del d_list[6:8]
print(d_list)
print(c_list)

# .copy()和赋值 = 不同
e_list = d_list
del e_list[1:4]
print(d_list)
print(e_list)

# .extend()在列表后再加个列表
print()
print(a_list)
a_list.extend(c_list)   #相当于 a_list += c_list
print(a_list)

# .insert(x,y) 在索引x处增加y

# 排序
# a_list.sort(),因为元素类型不一，所以报错

print()
print(a_list)
x = a_list.reverse()
print(a_list)
print(x) #reverse()只对当前序列操作并不返回一个可逆序列：返回值是 None
```

```python
import random
n = 3
a_list = [random.randrange(65, 91) for i in range(3)]
print(a_list)

# 插入
print(".insert()插入：")
a_list.insert(1, 'example')

# 删除
print()
print(a_list)
a_list.remove('example')
print(a_list)

# pop()删除，并且返回删除值

print()
a = a_list.pop(2)
print(a_list)
print(a)
```
